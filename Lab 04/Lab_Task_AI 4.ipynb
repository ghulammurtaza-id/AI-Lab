```python
from queue import PriorityQueue

class GoalBasedAgent:
    def __init__(self, graph, start, goal, limit):
        self.graph = graph
        self.start = start
        self.goal = goal
        self.limit = limit

    def depth_limited_search(self):
        stack = [(self.start, [self.start], 0)]
        while stack:
            node, path, depth = stack.pop()
            if node == self.goal:
                return path
            if depth < self.limit:
                for neighbor in self.graph.get(node, []):
                    if neighbor not in path:
                        stack.append((neighbor, path + [neighbor], depth + 1))
        return None

class UtilityBasedAgent:
    def __init__(self, graph, costs, start, goal):
        self.graph = graph
        self.costs = costs
        self.start = start
        self.goal = goal

    def uniform_cost_search(self):
        pq = PriorityQueue()
        pq.put((0, [self.start]))
        visited = set()
        while not pq.empty():
            cost, path = pq.get()
            node = path[-1]
            if node == self.goal:
                return path, cost
            if node not in visited:
                visited.add(node)
                for neighbor in self.graph.get(node, []):
                    if neighbor not in visited:
                        total_cost = cost + self.costs.get((node, neighbor), 1)
                        pq.put((total_cost, path + [neighbor]))
        return None, float('inf')

graph = {'A':['B','C'], 'B':['D','E'], 'C':['F'], 'D':[], 'E':['F'], 'F':[]}
costs = {('A','B'):2, ('A','C'):5, ('B','D'):1, ('B','E'):3, ('C','F'):2, ('E','F'):1}

goal_agent = GoalBasedAgent(graph,'A','F',4)
print("DLS Path:", goal_agent.depth_limited_search())

util_agent = UtilityBasedAgent(graph,costs,'A','F')
path,cost = util_agent.uniform_cost_search()
print("UCS Path:", path,"Cost:",cost)
```

```python
import itertools

cities = ['A','B','C','D']
distances = {
    ('A','B'):10, ('A','C'):15, ('A','D'):20,
    ('B','A'):10, ('B','C'):35, ('B','D'):25,
    ('C','A'):15, ('C','B'):35, ('C','D'):30,
    ('D','A'):20, ('D','B'):25, ('D','C'):30
}

shortest_path = None
min_cost = float('inf')

for perm in itertools.permutations(cities):
    cost = 0
    for i in range(len(perm)):
        cost += distances[(perm[i], perm[(i+1)%len(cities)])]
    if cost < min_cost:
        min_cost = cost
        shortest_path = perm

print("TSP Shortest Path:", shortest_path)
print("Total Distance:", min_cost)
```

```python
graph = {'A':['B','C'],'B':['D','E'],'C':['F'],'D':[],'E':['F'],'F':[]}

def dls(node, goal, limit, path):
    if node == goal:
        return path
    if limit <= 0:
        return None
    for neighbor in graph.get(node,[]):
        if neighbor not in path:
            res = dls(neighbor, goal, limit-1, path+[neighbor])
            if res:
                return res
    return None

def iddfs(start, goal, max_depth):
    for depth in range(max_depth+1):
        path = dls(start, goal, depth, [start])
        if path:
            return path
    return None

print("IDDFS Path:", iddfs('A','F',4))
```
