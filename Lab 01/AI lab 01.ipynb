```python
#task 3
import math
lst= list(map(int,input().split()))
query = int(input("which input you want to update index: "))
value = int(input("with what value "))

sum = 0;
mx = -math.inf
mn = math.inf

for i in range(len(lst)):
    print(lst[i],end=" ")
    sum+=lst[i]
    if lst[i]> mx:
        mx = lst[i]
    if lst[i]<mn:
        mn = lst[i]
    if i == query:
        lst[i] = value

print("updated list ")
print(lst)
print("this is max ",mx)
print("this is min ",mn)
```

```python
#task 4
def workk(amount):
    if amount<1000:
        return amount - amount*(5/100)
    elif amount<5000:
        return amount - amount*(10/100)
    else:
        return amount - amount*(15/100)

print(workk(500))
print(workk(1050))
print(workk(60000))
        
```

```python
#task 5
def dit_task(s):
    d = {
        "vowels" : 0,
        "consonants" : 0,
        "total" :len(s)
    }
    for c in s:
        if c=='o' or c=='O' or c=='A' or c=='a' or c=='e' or c=='E' or c=='e' or c=='I' or c=='i' or c=='u' or c=='U':
            d["vowels"] += 1
        else:
            d["consonants"] +=1
    return d

dowork = dit_task("Artificial Intellegnce")
for key,val in dowork.items():
    print(key," ",val)
    
```

```python
def work(lst):
    d = {}
    l = []
    
    for n in lst:
        if n in d:
            d[n] += 1
        else:
            d[n] = 1

    for num, cnt in d.items():
        if cnt > 1:
            l.append(num)
    
    return l

lst = list(map(int, input().split()))
ans = work(lst)
print(ans)
```
