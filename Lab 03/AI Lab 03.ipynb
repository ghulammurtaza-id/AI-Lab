```python
import random

class SecuritySystem:
    def __init__(self):
        self.components = {chr(i): random.choice([True, False]) for i in range(65, 74)}
       

    def show_state(self, title):
        print(f"\n{title}")
        for comp, state in self.components.items():
            status = "Safe" if state else "Vulnerable"
            print(f"{comp} : {status}")

    def patch(self, comp):
        self.components[comp] = True


class SecurityAgent:
    def __init__(self):
        self.memory = []

    def scan(self, system):
        print("\nAgent scanning the system...\n")
        for comp, state in system.components.items():
            if not state:
                print(f"Alert: {comp} is vulnerable.")
                self.memory.append(comp)
            else:
                print(f"{comp} checked. No issues found.")

    def repair(self, system):
        print("\nAgent repairing vulnerable components...\n")
        for comp in self.memory:
            system.patch(comp)
            print(f"{comp} patched successfully.")

    def verify(self, system):
        print("\nVerification after patching...\n")
        for comp, state in system.components.items():
            status = "Safe" if state else "Vulnerable"
            print(f"{comp} : {status}")



system = SecuritySystem()


system.show_state("Initial System State")

# Agent work
agent = SecurityAgent()
agent.scan(system)
agent.repair(system)
agent.verify(system)
```

```python
import random

class DataCenter:
    def __init__(self):
        self.servers = {
            f"Server-{i+1}": random.choice(["Underloaded", "Balanced", "Overloaded"])
            for i in range(5)
        }

    def show_status(self, title):
        print(f"\n{title}")
        for name, load in self.servers.items():
            print(f"{name} : {load}")

    def update_load(self, server, new_state):
        self.servers[server] = new_state


class LoadBalancerAgent:
    def __init__(self):
        self.overloaded = []
        self.underloaded = []

    def scan(self, datacenter):
        print("\nAgent scanning server loads...\n")
        for server, load in datacenter.servers.items():
            if load == "Overloaded":
                print(f"{server} is overloaded.")
                self.overloaded.append(server)
            elif load == "Underloaded":
                print(f"{server} is underloaded.")
                self.underloaded.append(server)
            else:
                print(f"{server} is already balanced.")

    def balance(self, datacenter):
        print("\nBalancing load between servers...\n")
        while self.overloaded and self.underloaded:
            heavy = self.overloaded.pop(0)
            light = self.underloaded.pop(0)

            datacenter.update_load(heavy, "Balanced")
            datacenter.update_load(light, "Balanced")

            print(f"Tasks moved from {heavy} to {light}.")

    def verify(self, datacenter):
        print("\nFinal server load status:\n")
        for server, load in datacenter.servers.items():
            print(f"{server} : {load}")



dc = DataCenter()
dc.show_status("Initial Server Load Status")

# Agent actions
agent = LoadBalancerAgent()
agent.scan(dc)
agent.balance(dc)
agent.verify(dc)
```

```python
import random

class BackupEnvironment:
    def __init__(self):
        self.tasks = {
            f"Backup-{i+1}": random.choice(["Completed", "Failed"])
            for i in range(8)
        }

    def show_tasks(self, heading):
        print(f"\n{heading}")
        for task, status in self.tasks.items():
            print(f"{task} : {status}")

    def retry_task(self, task):
        # After retry, assume success
        self.tasks[task] = "Completed"


class BackupManagementAgent:
    def __init__(self):
        self.failed_tasks = []

    def scan(self, environment):
        print("\nAgent scanning backup results...\n")
        for task, status in environment.tasks.items():
            if status == "Failed":
                print(f"{task} failed. Marked for retry.")
                self.failed_tasks.append(task)
            else:
                print(f"{task} completed successfully.")

    def retry_failures(self, environment):
        print("\nRetrying failed backups...\n")
        for task in self.failed_tasks:
            environment.retry_task(task)
            print(f"{task} retried and completed.")

    def verify(self, environment):
        print("\nUpdated backup task statuses:\n")
        for task, status in environment.tasks.items():
            print(f"{task} : {status}")



backup_system = BackupEnvironment()
backup_system.show_tasks("Initial Backup Task Status")


agent = BackupManagementAgent()
agent.scan(backup_system)
agent.retry_failures(backup_system)
agent.verify(backup_system)
```

```python
import random

class SecurityEnvironment:
    def __init__(self):
        states = ["Safe", "Low Risk Vulnerable", "High Risk Vulnerable"]
        self.components = {chr(i): random.choice(states) for i in range(65, 74)}

    def display(self, title):
        print(f"\n{title}")
        for comp, state in self.components.items():
            print(f"{comp} : {state}")

    def patch_low_risk(self, comp):
        self.components[comp] = "Safe"


class UtilitySecurityAgent:
    def __init__(self):
        self.low_risk_targets = []
        self.high_risk_targets = []

    def evaluate(self, env):
        print("\nScanning components and evaluating risk...\n")
        for comp, state in env.components.items():
            if state == "Low Risk Vulnerable":
                print(f"{comp} has low risk vulnerability. Suitable for patching.")
                self.low_risk_targets.append(comp)
            elif state == "High Risk Vulnerable":
                print(f"{comp} has HIGH risk vulnerability. Premium service required.")
                self.high_risk_targets.append(comp)
            else:
                print(f"{comp} is safe.")

    def act(self, env):
        print("\nApplying available security service...\n")
        for comp in self.low_risk_targets:
            env.patch_low_risk(comp)
            print(f"{comp} patched using basic service.")

        for comp in self.high_risk_targets:
            print(f"{comp} left unpatched (requires premium service).")

    def report(self, env):
        print("\nSystem status after agent action:\n")
        for comp, state in env.components.items():
            print(f"{comp} : {state}")



environment = SecurityEnvironment()
environment.display("Initial System State")


agent = UtilitySecurityAgent()
agent.evaluate(environment)
agent.act(environment)
agent.report(environment)
```

```python
class HospitalMap:
    def __init__(self):
        # room -> (patient_id, medicine, priority) ky liye
        self.tasks = {
            "Ward-A": ("PX21", "Insulin", "High"),
            "Ward-B": ("PX34", "Antibiotic", "Medium"),
            "ICU-1": ("PX09", "Painkiller", "High"),
            "Room-12": ("PX77", "Vitamin", "Low")
        }

        self.available_medicines = ["Insulin", "Antibiotic", "Painkiller", "Vitamin"]

    def check_identity(self, room, pid):
        return self.tasks[room][0] == pid


class DeliveryRobot:
    def __init__(self, hospital):
        self.hospital = hospital
        self.position = "Docking-Station"
        self.load = None

    def move(self, destination):
        self.position = destination
        print(f"Robot reached {destination}")

    def collect_medicine(self, med):
        if med in self.hospital.available_medicines:
            self.load = med
            print(f"Collected {med}")

    def verify_and_deliver(self, room, pid):
        print(f"Verifying patient in {room}")
        if self.hospital.check_identity(room, pid):
            print(f"Identity confirmed. Delivered {self.load} to {room}")
            self.load = None
        else:
            print("Identity mismatch. Calling nurse.")
            self.call_nurse(room)

    def call_nurse(self, room):
        print(f"Nurse alerted to {room}")

    def execute_goal(self):
        print("\nRobot executing delivery plan...\n")
        for room, info in self.hospital.tasks.items():
            pid, med, priority = info

            print(f"\nTask → {room} | Priority: {priority}")

            self.move("Pharmacy")
            self.collect_medicine(med)

            self.move(room)
            self.verify_and_deliver(room, pid)


hospital = HospitalMap()
robot = DeliveryRobot(hospital)
robot.execute_goal()
```

```python

class Building:
    def __init__(self):
       
        self.rooms = {
            'a': 'safe', 'b': 'safe', 'c': 'fire',
            'd': 'safe', 'e': 'fire', 'f': 'safe',
            'g': 'safe', 'h': 'safe', 'j': 'fire'
        }

        
        self.grid = [
            ['a', 'b', 'c'],
            ['d', 'e', 'f'],
            ['g', 'h', 'j']
        ]

    
    def display(self):
        for row in self.grid:
            for room in row:
                if self.rooms[room] == 'fire':
                    print("[fire]", end=" ")
                else:
                    print("[ ]", end=" ")
            print()
        print()

    # Function to extinguish fire
    def put_out_fire(self, room):
        self.rooms[room] = 'safe'



class Robot:
    def __init__(self, building):
        self.building = building
        self.path = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'j']

   
    def move(self):
        print("Initial Building Condition:")
        self.building.display()

        for room in self.path:
            print(f"Robot moves to room {room}")

            if self.building.rooms[room] == 'fire':
                print("Fire found! Extinguishing fire...")
                self.building.put_out_fire(room)
            else:
                print("No fire in this room.")

            print("Current Building Condition:")
            self.building.display()

        print("All rooms visited. Final Building Condition:")
        self.building.display()


building = Building()
robot = Robot(building)
robot.move()
```
